```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np

import swotdawgviz
import swotdawgviz.io as sdvio
import swotdawgviz.plots as sdvpl
```

## Set paths

```python
outputdir = "FILL_THIS" # Path to directory containing the H2iVDI output nc files
sosdir = "FILL_THIS" # Path to directory containing the SoS nc files
swotncdir = "FILL_THIS" # Path to directory containing the SWOT nc observations files
```

## Set ID of the reach to analyse

```python
reach_id = 74269500051
```

## Load model mean discharge from SoS

```python
sos = sdvio.SosNetCDF(os.path.join(sosdir, "na_sword_v11_SOS_priors.nc"), verbose=False)
reach_data = sos.get_reach(reach_id)
model_mean_q = float(reach_data["model_mean_q"])
```

## Load H2iVDI output

```python
h2ivdi = sdvio.OutputH2iVDI(os.path.join(outputdir, "%i_hivdi.nc" % reach_id))
```

## Create plot using matplotlib backend

```python
qplot = sdvpl.DischargePlot(title="Reach %i, status:%i" % (reach_id, h2ivdi.status()))
qplot.add_prior("prior mean Q", model_mean_q, color="r", linestyle="--")
qplot.add_product("H2iVDI", h2ivdi.Q, times=h2ivdi.t,
                  color="b", linestyle="-")
qplot.render()
```

## Create plot using plotly backend

```python
qplot = sdvpl.DischargePlot(title="Reach %i, status:%i" % (reach_id, h2ivdi.status()))
qplot.add_prior("prior mean Q", model_mean_q, color="red", linestyle="dash")
qplot.add_product("H2iVDI", h2ivdi.Q, times=h2ivdi.t,
                  color="blue", linestyle=None)
qplot.render(backend="plotly")
```
