```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import hydroviz
import hydroviz.io as sdvio
import hydroviz.plots as sdvpl
```

## Set paths

```python
outputdir = "/mnt/data/output" # Path to directory containing H2iVDI output nc files
sosdir = "/mnt/data/input/sos" # Path to directory containing SoS nc files
```

## Set ID of the reach to analyse

```python
reach_id = 23214400041
```

## Load model mean discharge from SoS

```python
sos = sdvio.SosNetCDF(os.path.join(sosdir, "eu_sword_reaches_hb23_v16_SOS_priors.nc"), verbose=False)
reach_data = sos.get_reach(reach_id)
model_mean_q = float(reach_data["model_mean_q"])
```

## Load insitu discharge (O2620020)

```python
insitu_dir = "/mnt/data/insitu_data"
insitu_data = pd.read_csv(os.path.join(insitu_dir, "O2620020_QmnJ(n=1_non-glissant).csv"), sep=",", parse_dates=[0])
```

```python
insitu_data.iloc[:, 0]
```

## Load H2iVDI output

```python
h2ivdi = sdvio.OutputH2iVDI(os.path.join(outputdir, "%i_h2ivdi.nc" % reach_id))
```

## Create plot using matplotlib backend

```python
qplot = sdvpl.DischargePlot(title="Reach %i, status:%i" % (reach_id, h2ivdi.status()), date_units="datetime64")
qplot.add_prior("prior mean Q", model_mean_q, color="r", linestyle="--")
qplot.add_product("H2iVDI", h2ivdi.Q, times=h2ivdi.dates,
                  color="b", linestyle="-")
print(h2ivdi.dates.dtype, insitu_data.iloc[:, 0].dtype)
qplot.add_product("in-situ", insitu_data.iloc[:, 1], times=insitu_data.iloc[:, 0],
                  color="r", linestyle="dotted")
qplot.set_dates_xaxis()
qplot.render()
```

## Create plot using plotly backend

```python
qplot = sdvpl.DischargePlot(title="Reach %i, status:%i" % (reach_id, h2ivdi.status()))
qplot.add_prior("prior mean Q", model_mean_q, color="red", linestyle="dash")
qplot.add_product("H2iVDI", h2ivdi.Q, times=h2ivdi.t,
                  color="blue", linestyle=None)
qplot.render(backend="plotly")
```
