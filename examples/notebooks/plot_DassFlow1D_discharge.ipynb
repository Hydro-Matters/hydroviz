```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import hydroviz
import hydroviz.io as hvio
import hydroviz.plots as sdvpl
```

## Set paths

```python
out_file = "../data/dassflow1d/dassflow1d_out.nc"
xs_shp = "../data/dassflow1d/xs_nodes.shp"
```

## Load DassFlow1D output

```python
dassflow1d = hvio.OutputDassFlow1D(out_file)
```

## Load insitu discharge (O9090010)

```python
insitu_dir = "../data/insitu_data"
insitu_data = pd.read_csv(os.path.join(insitu_dir, "O9090010_QmnH(n=3_non-glissant).csv"), sep=",", parse_dates=[0])
```

## Create plot using matplotlib backend

```python
qplot = sdvpl.DischargePlot(title="Station Marmande (O9090010)", date_units="datetime64")
#qplot.add_prior("prior mean Q", model_mean_q, color="r", linestyle="--")
qplot.add_product("DassFlow1D", dassflow1d.Q[:, 41], times=dassflow1d.dates,
                  color="b", linestyle="-")
#print(h2ivdi.dates.dtype, insitu_data.iloc[:, 0].dtype)
qplot.add_product("in-situ", insitu_data.iloc[:, 1], times=insitu_data.iloc[:, 0],
                  color="r", linestyle="dotted")
qplot.set_dates_xaxis()
qplot.render()
```

## Create plot using plotly backend

```python
qplot = sdvpl.DischargePlot(title=None)
qplot.add_product("in-situ", insitu_data.iloc[:, 1], times=insitu_data.iloc[:, 0], color="red", linestyle="dash")
qplot.add_product("DassFlow1D", dassflow1d.Q[:, 41], times=dassflow1d.dates,
                  color="blue", linestyle=None)
qplot.render(backend="plotly")
```
