```python
import hydroviz
import hydroviz.io as hvio
import hydroviz.maps as hvm
```

## Set paths

```python
out_file = "../data/MGB/mgb_out.nc"
mgb_poly_shp = "../data/MGB/Maroni_MGB_Catchments.shp"
mgb_riv_shp = "../data/MGB/Maroni_MGB_Drainage.shp"
```

## Load MGB output

```python
mgb = hvio.OutputMGB(out_file, mgb_poly_shp, mgb_riv_shp)
```

## Create map of discharge

```python
cmap = hvm.CatchmentsMap(mgb, backend="folium")
m = cmap.get_catchments_map("discharge")
m
```
