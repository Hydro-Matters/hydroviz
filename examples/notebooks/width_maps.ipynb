## Set paths to the directories containing SWORD and SWOT observations files

```python
sword_shp_dir = "FILL_THIS"
swot_nc_dir = "FILL_THIS"
```

## Load collection of SWOT observations files for reaches

```python
import os
import swotdawgviz.io as sdvio
```

```python
collection = sdvio.SwotObservationsCollection(swot_nc_dir)
```

```python
collection.reaches_list
```

## Load SWORD data for reaches in collection

```python
sword_hb74_reaches = sdvio.SwordShapefile(os.path.join(sword_shp_dir, "na_sword_reaches_hb74_v1.shp"),
                                          reaches_list=collection.reaches_list)
```

## Plot maps of reaches colored with the width attribute

```python
import swotdawgviz.maps as sdvm
```

```python
rmap = sdvm.ReachesMap(sword_hb74_reaches.dataset)
```

### Plot colored centerlines of reaches

```python
m = rmap.get_centerlines_map("width")
m
```

### Plot colored polygons of reaches

```python
m = rmap.get_polygons_map(varname="width", width_attribute="width")
m
```
