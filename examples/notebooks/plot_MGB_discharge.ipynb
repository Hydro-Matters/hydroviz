```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import hydroviz
import hydroviz.io as hvio
import hydroviz.plots as sdvpl
```

## Set paths

```python
out_freerun_file = "../data/MGB/mgb_out_freerun.nc"
out_assim_file = "../data/MGB/mgb_out_assim.nc"
```

## Load MGB (FreeR output

```python
mgb_freerun = hvio.OutputMGB(out_freerun_file)
mgb_assim = hvio.OutputMGB(out_assim_file)
```

## Load insitu discharge (O9090010)

```python
insitu_dir = "../data/insitu_data"
insitu_data = pd.read_csv(os.path.join(insitu_dir, "52410001_QmnJ(n=1_non-glissant).csv"), sep=",", parse_dates=[0])
```

## Create plot using matplotlib backend

```python
qplot = sdvpl.DischargePlot(title="Station Grand-Santi (52410001)", date_units="datetime64")
#qplot.add_prior("prior mean Q", model_mean_q, color="r", linestyle="--")
qplot.add_product("MGB (Free-run)", mgb_freerun.Q[:, 4409], times=mgb_freerun.dates,
                  color="b", linestyle="-")
qplot.add_product("MGB (Assimilation)", mgb_assim.Q[:, 4409], times=mgb_assim.dates, unc=mgb_assim.Q_u[:, 4409],
                  color="g", linestyle="--")
#print(h2ivdi.dates.dtype, insitu_data.iloc[:, 0].dtype)
qplot.add_product("in-situ", insitu_data.iloc[:, 1], times=insitu_data.iloc[:, 0],
                  color="r", linestyle="dotted")
qplot.set_dates_xaxis()
qplot.render()
```
