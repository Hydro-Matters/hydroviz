```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import hydroviz
import hydroviz.io as hvio
import hydroviz.maps as hvm
```

## Set paths

```python
inputdir = "../data/h2ivdi/input" # Path to directory containing input files and subdirectories
outputdir = "../data/h2ivdi/output" # Path to directory containing H2iVDI output nc files
setsfile = "sets.json" # File containing sets definition
sword_shp_dir = "../data/SWORD/V16/shp/" # Path to the directory that contains continents subdirectories of SWORD shapefiles
```

## Load H2iVDI results

```python
results = hvio.DischargeAlgorithmResults("h2ivdi", inputdir, outputdir, setsfile, sword_shp_dir=sword_shp_dir, basinID=616, simplify_tolerance=0.001)
```

## Create map of discharge

```python
rmap = hvm.ReachesMap(results, tiles="GoogleHybrid")
m = rmap.get_centerlines_map("discharge")
m
```

## Create map of monthly mean discharge

```python
rmap = hvm.ReachesMap(results, tiles="GoogleHybrid")
m = rmap.get_centerlines_map("discharge", temporal_mean="M")
m
```
